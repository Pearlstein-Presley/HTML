```python
# pyramid of stars
n=5
for i in range(1, n + 1):
  print(" " * (n-i) + "* " * i)
```

```python
# Inverted stars
n = 5
for i in range(n, 0, -1):
  print("*"*i)
```

```python
# Floyd triangle
n = 5
num = 1

for i in range(1, n + 1):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
```

```python
# pascal triangle
n = 5

for i in range(n):
    num = 1
    print(" " * (n - i), end="")

    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)

    print()
```

```python
#diamond pattern using stars
n = 5

for i in range(1, n + 1):
    print(" " * (n - i) + "* " * i)

for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "* " * i)
```

```python
#reverse string
s = "hello"
rev = ""
for ch in s:
  rev = ch + rev
  print(rev)
```

```python
#string is a palindrome
s = "madam"
rev = ""
for ch in s:
  rev = ch +rev
  if s == rev:
    print("Pallindrome")
  else:
      print("Not a Pallindrome")
```

```python
#Frequency of stars
s = "hello"
for ch in s:
  if s.index(ch) == s.find(ch):
    print(ch, ":", s.count(ch))
```

```python
#Remove vowels from strings

s = "education"
result =""
for ch in s:
  if ch not in "aeiouAEIOU":
    result += ch
    print(result)
```

```python
#two strings are anagrams

s1 = "listen"
s2 = "slient"
if sorted(s1) == sorted(s2):
  print("Anagrams")
else:
    print("Not anagrams")
```

```python
# largest and smallest
a = [10, 25, 5, 40, 15]

largest = a[0]
smallest = a[0]

for n in a:
    if n > largest:
        largest = n
    if n < smallest:
        smallest = n

print("Largest:", largest)
print("Smallest:", smallest)
```

```python
#duplicates form array
a = [1 ,2 ,3 ,4 ,5]
result =[]
for n in a:
  if n not in result:
    result.append(n)
    print(result)
```

```python
#rotate an array by k positions
a = [1 ,2 ,3 ,4 ,5]
k = 2
k = k % len(a)
result = a[-k:] + a[:-k]
print(result)
```

```python
#find the number of missing sequence
a = [1 ,2 ,3 ,4 ,5]
n = 6
total = n * (n + 1) // 2
missing = total - sum(a)
print("Missing number:", missing)
```

```python
# Armstrong number
n = 153
temp = n
total = 0
power = len(str(n))
while temp > 0:
  digit = temp % 10
  total += digit ** power
  temp //= 10
  if total == n:
    print("Armstrong number")
  else:
      print("Not an armstrong number")
```

```python
#check whether a number is a perfetc number
n = 6
total = 0

for i in range(1, n):
    if n % i == 0:
        total += i

if total == n:
    print("Perfect number")
else:
    print("Not a perfect number")
```

```python
#find gcd and lcm of two numbers
a = 12
b = 18

x = a
y = b

while y != 0:
    x, y = y, x % y

gcd = x
lcm = (a * b) // gcd

print("GCD:", gcd)
print("LCM:", lcm)
```

```python
#Print all prime numbers in a range
start = 1
end = 20

for n in range(start, end + 1):
    if n > 1:
        prime = True

        for i in range(2, n):
            if n % i == 0:
                prime = False
                break

        if prime:
            print(n, end=" ")
```

```python
#convert decimal to binary without using bin()

n = 10
binary = ""

if n == 0:
  binary = "0"

while n > 0:
  remainder = n % 2
  binary = str(remainder) + binary
  n = n // 2
print("Binary:", binary)
```
