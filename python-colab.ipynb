Data Cleaning

```python
import pandas as pd
import numpy as np
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
df = pd.read_csv('Salary.csv')
```

```python
df.info()
```

```python
df.head(20)
```

```python
# Check empty cells

print(df.isna().any())
```

```python
# Count non null values

print(df.count())
```

```python
# check column data types

print(df.dtypes)
```

```python
# convert column names to lower case

df.columns = df.columns.str.lower()
print(df.columns)
```

```python
# remove extra space from columns

df.columns = df.columns.str.strip()

print(df.columns)
```

```python
# check infinte values

print(np.isinf(df.select_dtypes(include='number')).sum())
```

```python
# check uniq values

print(df['yearsexperience'].unique())
```

```python
# check uniq values for salary

print(df['salary'].unique())
```

```python
# check salary distribution

print(df['salary'].value_counts())
```

```python
 # check experince distribution

 print(df['yearsexperience'].value_counts())
```

```python
# check duplicates records

print(df['yearsexperience'].duplicated().sum())
```

```python
# checking salary greater than average

average_salary = df['salary'].mean()
print(df[df['salary'] > average_salary])
```

```python
# check missing values after cleaning

print(df.isna().sum())
```

```python
# removing empty rows

df = df.dropna(how='all')
print(df)
```

```python
# Removing empty columns

df = df.dropna(axis=1, how='all')
print(df)
```

```python
# Exporting the final dataset

df.to_csv('Salary_Final_Cleaned.csv', index=False)
print("Dataset cleaned and saved successfully!")
```
