```python
# Arithmetic Operator Example

a=10
b=5
print("Addition:",a+b)
print("Sbutraction:",a-b)
print("Multiplication:",a*b)
print("Division:",a/b)
```

```python
# Relational Operator Example

a=10
b=5
print("Equal to:", a==b)
print("Greater than:",a>b)
print("Lesser than:",a<b)
```

```python
# Logical Operators Example

a=10
b=5
print("AND Operator:",a>5 and b<10)
print("OR Operator:",a<5 or b>10)
print("NOT Operator:",not(a<5))
```

```python
# Assignment Operators Example

a=10
print("Assign value:",a)

a+=10
print("Add And Assign:",a)

a-=10
print("Subtract And Assign:",a)

a*=10
print("Multiply And Assign:",a)

a/=10
print("Division And Assign:",a)

a=10
a%=10
print("Modulus And Assign:",a)
```

```python
# Bitwise Operator Example

a=10
b=5
print("Bitwise AND:",a&b)
print("Bitwise OR:",a|b)
print("Bitwise XOR:",a^b)
print("Bitwise NOT:",-a)
print("Left Shift:",a<< 1)
print("Right Shift:",a<< 1)
```
